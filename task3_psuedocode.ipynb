### Q1. Find the Frequency of Each Word in a Sentence

### 1. Pseudocode

```
START
INPUT a sentence
SPLIT the sentence into words
CREATE an empty dictionary
FOR each word in the sentence
    COUNT how many times the word appears
    STORE the word and its count in the dictionary
END FOR
RETURN the dictionary
STOP
```



### Expected Output

```
{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}
{'hello': 2, 'world': 1}
```


````python
def q1_word_frequency(s):
    w = s.split()
    d = {}
    for i in w:
        d[i] = w.count(i)
    return d

print(q1_word_frequency("python is easy and python is powerful"))
print(q1_word_frequency("hello world hello"))
````

## Q3. Find the First Non-Repeating Character in a Word

### Pseudocode

```
START
INPUT a string
FOR each character in the string
    COUNT how many times the character appears
    IF the count equals 1
        RETURN the character
    END IF
END FOR
IF no non-repeating character is found
    RETURN nothing
STOP
```

### Expected Output

Input: `swiss`

```
w
```



````python
def q3_First_non_repeating_char(s):
    d = {}
    for i in s:
        d[i]=s.count(i)
    for i in d:
        if d[i] == 1:
            return i
    
print(q3_First_non_repeating_char('swiss'))
````

## Q6. Move All Zeros to the End of a List

### Pseudocode

```
START
INPUT a list
CREATE an empty list
FOR each element in the list
    IF the element is not zero
        ADD the element to the new list
    END IF
END FOR
COUNT the number of zeros in the original list
ADD all zeros to the end of the new list
RETURN the new list
STOP
```

### Expected Output

Input: `[0, 3, 0, 1, 2, 0, 4]`

```
[3, 1, 2, 4, 0, 0, 0]
```


````python
def q6_All_Zeros(l):
    l1 = []
    for i in l:
        if i != 0:
            l1.append(i)
    n = l.count(0)
    return l1+[0]*n

print(q6_All_Zeros([0, 3, 0, 1, 2, 0, 4]))
````

## Q7. Find the Second Highest Number in a List

### Pseudocode

```
START
INPUT a list
CREATE an empty list
FOR each number in the original list
    IF the number is not in the new list
        ADD the number to the new list
    END IF
END FOR
SORT the new list in ascending order
IF the list contains fewer than two numbers
    RETURN nothing
ELSE
    RETURN the second-last number
END IF
STOP
```

### Expected Output

Input: `[10, 20, 40, 30, 40, 20]`

```
30
```



````python
def q7_Second_Highest(l):
    l1 = []
    for i in l:
        if i not in l1:
            l1.append(i)
    l1.sort()
    return l1[-2]


print(q7_Second_Highest([10, 20, 40, 30, 40, 20]))
````

## Q9. Find the Missing Number from 1 to N

### Pseudocode

```
START
INPUT a list and N
FOR each number from 1 to N
    IF the number is not present in the list
        RETURN the number
    END IF
END FOR
RETURN nothing
STOP
```

### Expected Output

Input: `[1, 2, 4, 5, 6]`, N = `6`

```
3
```


````python
def q9_Missing_Number(l,n):
    l1 = []
    for i in range(1,n+1):
        l1.append(i)
    for i in l1:
        if i not in l:
            return i

print(q9_Missing_Number([1,2,4,5,6],6))
````

## Q10. Find Two Numbers Whose Sum Equals a Target

### Pseudocode

```
START
INPUT a list and target
CREATE an empty dictionary
FOR each number in the list
    CALCULATE complement = target - number
    IF complement exists in the dictionary
        RETURN complement and number
    END IF
    STORE the number in the dictionary
END FOR
RETURN an empty list
STOP
```

### Expected Output

Input: `[2, 7, 11, 15]`, Target = `9`

```
[2, 7]
```


````python
def q10_two_sum(nums, target):
    seen = {}
    for num in nums:
        complement = target - num
        if complement in seen:
            return [complement, num]
        seen[num] = True
    return []


print(q10_two_sum([2, 7, 11, 15],9))
````

## Q11. Find the Number Closest to Zero

### Pseudocode

```
START
INPUT a list of numbers
SET the first number as the closest number
FOR each number in the list
    IF its absolute value is smaller than the closest number's absolute value
        UPDATE the closest number
    ELSE IF both absolute values are equal
        SELECT the positive number
    END IF
END FOR
RETURN the closest number
STOP
```

### Expected Output

Input: `[-8, -3, 2, 5, -2]`

```
2
```


````python
def q11_Number_Closest(l):
    l1 = []
    l2 = []
    for i in l:
        if i >= 0:
            l1.append(i)
        else:
            l2.append(i)
    l1.sort()
    l2.sort()
    if l2[0]*-1 >= l1[0]:
        return l1[0]
    else:
        return l2[0]
        
print(q11_Number_Closest([-8, -3, 2, 5, -2]))
````

## Q12. Find Common Elements Between Two Lists

### Pseudocode

```
START
INPUT two lists
CREATE an empty list
FOR each element in the first list
    IF the element exists in the second list
        ADD the element to the new list
    END IF
END FOR
RETURN the new list
STOP
```

### Expected Output

Input: `[1, 2, 3, 4, 5]`, `[3, 4, 5, 6, 7]`

```
[3, 4, 5]
```


````python
def q12_commom_elements(l1,l2):
    l3=[x for x in l1 if x in l2]
    return l3

print(q12_commom_elements([1, 2, 3, 4, 5],[3, 4, 5, 6, 7]))
````

## Q13. Rotate a List to the Right by K Positions

### Pseudocode

```
START
INPUT a list and K
IF the list is empty
    RETURN the empty list
END IF
CALCULATE K modulo the length of the list
TAKE the last K elements
PLACE them at the beginning of the list
PLACE the remaining elements after them
RETURN the rotated list
STOP
```

### Expected Output

Input: `[1, 2, 3, 4, 5]`, K = `2`

```
[4, 5, 1, 2, 3]
```



````python
def q13_K_Positions(l,k):
    l1 = []
    for i in range(1,k+1):
        l1.append(l[i*-1])
    l1 = l1[::-1]
    for i in l:
        if i not in l1:
            l1.append(i)
    return l1

print(q13_K_Positions( [1, 2, 3, 4, 5],2))
````

## Q14. Find the Maximum Sum of Two Consecutive Elements

### Pseudocode

```
START
INPUT a list
CREATE an empty list for sums
FOR each element except the last element
    ADD the current element and the next element
    STORE their sum in the new list
END FOR
FIND the maximum sum
RETURN the maximum sum
STOP
```

### Expected Output

Input: `[4, 7, 2, 9, 5]`

```
14
```


````python
def q14_max_sum_of_consecutive(l):
    l1= []
    for i in range(len(l)-1):
        l1.append(l[i]+l[i+1])
    return(max(l1))
print(q14_max_sum_of_consecutive([4, 7, 2, 9,5]))
````

## Q15. Find All Pairs of Numbers Whose Sum Equals a Target

### Pseudocode

```
START
INPUT a list and target
CONVERT the list into a set to remove duplicates
CREATE an empty list for pairs
FOR each number in the set
    CALCULATE complement = target - number
    IF complement exists in the set
       AND number is less than complement
        ADD the pair to the result list
    END IF
END FOR
SORT the pairs
RETURN the result list
STOP
```

### Expected Output

Input: `[2, 4, 3, 5, 7, 8, 1]`, Target = `9`

```
[[1, 8], [2, 7], [4, 5]]
```


````python
def q15_all_pairs(nums,target):
    l1 = []
    s = set(nums)
    for i in s:
        t = target-i
        if t in s and i < t:
            l1.append([i,t])
    return l1
print(q15_all_pairs( [2, 4, 3, 5, 7, 8, 1],9))
````
